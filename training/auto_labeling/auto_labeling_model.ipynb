```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")

model.train(
    data="/content/drive/MyDrive/yolo_dataset_trash/data.yaml",
    epochs=200,
    imgsz=640,
    batch=16,
    name="trash_auto_label"
)
```

```python
from ultralytics import YOLO

# =========================================================
# 1. 학습된 모델 불러오기
# =========================================================

MODEL_PATH = "/content/runs/detect/trash_auto_label/weights/best.pt"

model = YOLO(MODEL_PATH)


# =========================================================
# 2. Test 이미지 예측
# =========================================================

results = model.predict(
    source="/content/drive/MyDrive/yolo_dataset_trash/images/test",
    conf=0.5,
    save=True
)

print("Test 이미지 예측 완료!")
```

```python
from ultralytics import YOLO

# =========================================================
# 1. 학습된 YOLO 모델 불러오기
# =========================================================

MODEL_PATH = "/content/runs/detect/trash_auto_label/weights/best.pt"

model = YOLO(MODEL_PATH)


# =========================================================
# 2. Test 데이터로 성능 평가
# =========================================================

results = model.val(
    data="/content/drive/MyDrive/yolo_dataset_trash/data.yaml",
    split="test"
)


# =========================================================
# 3. 성능 출력
# =========================================================

print()
print("=" * 50)
print("Test 데이터 성능 평가 결과")
print("=" * 50)

print("Precision :", results.box.mp)
print("Recall    :", results.box.mr)
print("mAP50     :", results.box.map50)
print("mAP50-95  :", results.box.map)



```

```python
# =========================================================
# 4. 성능 평가 결과 TXT 파일로 저장
# =========================================================

result_text = f"""
==================================================
Test 데이터 성능 평가 결과
==================================================

Precision : {results.box.mp}
Recall    : {results.box.mr}
mAP50     : {results.box.map50}
mAP50-95  : {results.box.map}
"""

with open("/content/test_result.txt", "w", encoding="utf-8") as f:
    f.write(result_text)

print()
print("TXT 파일 저장 완료!")
print("저장 위치 : /content/test_result.txt")
```
